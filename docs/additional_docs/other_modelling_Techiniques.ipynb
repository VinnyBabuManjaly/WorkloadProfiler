## Alternative Clustering Algorithms

| Technique | Description | Suitability |
|-----------|-------------|-------------|
| **Gaussian Mixture Models (GMM)** | Probabilistic clustering assuming Gaussian distributions. Allows soft assignments (workloads can belong to multiple clusters with probabilities). | High – Handles overlapping workload types well |
| **K-Means** (various k values) | Partitioning into k spherical clusters. Faster than density methods for large datasets. | Medium – Already used as baseline; limited by spherical cluster assumption |
| **Agglomerative Hierarchical Clustering** | Bottom-up merging of clusters; produces a dendrogram showing relationships between workload types. | High – Useful for discovering hierarchical workload taxonomies |
| **BIRCH** | Balanced Iterative Reducing and Clustering using Hierarchies. Memory-efficient for large-scale data. | High – Designed for large datasets like this one |
| **Spectral Clustering** | Uses graph eigenvalues to find clusters in non-convex shapes. | Medium – Good for complex patterns but computationally expensive |
| **Mean Shift** | Finds dense regions without assuming cluster shapes or counts. | Low – Computationally intensive; may struggle with high-dimensional data |
| **OPTICS** | Similar to DBSCAN but orders points to handle varying densities better. | High – Could outperform DBSCAN on this dataset |

## Feature Representation Alternatives

| Technique | Description |
|-----------|-------------|
| **t-SNE / UMAP** | Non-linear dimensionality reduction for visualization and clustering. UMAP preserves global structure better than t-SNE. |
| **Autoencoders** | Neural network-based dimensionality reduction; learns compressed representations of workload features. |
| **Kernel PCA** | Non-linear PCA using kernel trick; captures more complex patterns than linear PCA. |

## Ensemble & Hybrid Approaches

| Approach | Description |
|----------|-------------|
| **Consensus Clustering** | Run multiple algorithms (K-Means, GMM, HDBSCAN) and combine results; improves stability |
| **Two-stage clustering** | First reduce with UMAP, then apply HDBSCAN on the embedding |
| **Subspace Clustering** | Clusters in different feature subspaces (e.g., CPU-heavy vs memory-heavy patterns simultaneously) |

## Key Recommendation

The README notes at line 98: *"Try alternative algorithms (GMM, DBSCAN), refine features"* as a contingency plan. **GMM** would be the strongest addition because:
- It provides probabilistic cluster membership (useful for workloads that might be hybrid types)
- Naturally handles clusters of different sizes and correlations
- Works well with the standardized features already prepared

The **UMAP + HDBSCAN** combination is also popular in modern clustering pipelines for better structure discovery than PCA alone.
